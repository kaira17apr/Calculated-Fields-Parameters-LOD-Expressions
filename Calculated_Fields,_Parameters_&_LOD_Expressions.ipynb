**Theoretical Questions :**

**Q1. What is a Calculated Field and why is it important in business dashboards?**

**Ans-**

A Calculated Field is a custom field created by applying formulas or expressions to existing data (e.g., Sales × Quantity = Total Revenue).

**Importance in Dashboards:**

*   Adds new insights beyond raw data
*   Helps tailor KPIs to business needs
*   Enables deeper analysis (like profit margins, growth rates)
*   Makes dashboards more dynamic and decision‑focused











**Q2. Why can’t parameters change visuals directly in Tableau?**

**Ans-**

In Tableau, parameters cannot change visuals directly because they are just placeholders for a single value (like a number, string, or date).

**To affect visuals, parameters must be combined with a calculated field or filter.**


*   **Parameter** → **Calculated Field/Filter → Visualization**
*   This design gives flexibility: one parameter can drive multiple calculations, filters, or reference lines, instead of being tied to visuals automatically.






**Q3. When should EXCLUDE LOD be used?**

**Ans-**

EXCLUDE LOD is used to ignore specific dimensions in a calculation so you can show broader context while still breaking down data by other fields.

**Typical use cases:**

*   When you want an overall value (like average sales across all regions) but still display it alongside each region.

*   To compare individual values against a broader aggregate (e.g., each product’s sales vs. overall average sales).
*   When you need to prevent certain dimensions from affecting the calculation.





**Q4. Difference between FIXED and INCLUDE LOD expressions?**


**Ans-**



*   **FIXED LOD** → Ignores the view and calculates at the exact dimension(s) you specify.

Example: {FIXED [Region]: SUM([Sales])} → Always gives sales per region, no matter what filters or dimensions are in the view.

*   **INCLUDE LOD** → Adds extra dimensions to the view’s context for calculation.

Example: {INCLUDE [Product]: AVG([Sales])} → Calculates average sales at product level, then rolls it up to whatever dimensions are in the view.
